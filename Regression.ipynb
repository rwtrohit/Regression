2).
What are the key assumptions of Simple Linear Regression ?

Ans2).
The key assumptions of simple linear regression are linearity, independence, homoscedasticity, and normality of residuals.

1. Linearity
The relationship between the independent variable (X) and the dependent variable (Y) is linear.

That means Y changes at a constant rate with respect to X.

2. Homoscedasticity
The variance of the errors (residuals) is constant across all levels of X.

If this is violated (called heteroscedasticity), predictions may be unreliable.

3. Independence
Observations are independent of each other.

No correlation between the residuals (errors); often violated in time-series data.

4. Normality of Errors
The residuals (differences between observed and predicted Y) are normally distributed, especially important for hypothesis testing and confidence intervals.


5. No (or little) multicollinearity (Not applicable to simple linear regression specifically, but crucial in multiple regression)
Since there's only one predictor in simple linear regression, this isn't relevant here.

6. No significant outliers or high leverage points
Extreme values can disproportionately influence the regression line and distort results.

5).
 How do we calculate the slope m in Simple Linear Regression ?


Ans5).
The slope (m) in simple linear regression is calculated using the formula: m = (n * Σxy - Σx * Σy) / (n * Σx² - (Σx)²). This formula utilizes the sums of the independent (x) and dependent (y) variable values, their products (xy), and the squares of the independent variable values (x²).
